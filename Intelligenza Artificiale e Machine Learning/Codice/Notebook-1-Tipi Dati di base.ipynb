```python
'this is an integer %d' % 15   # replacement di un int object
```

```python
'this is an integer %4d' % 15  # replacement con un numero fisso di caratteri
```

```python
'this is an integer %04d' % 15  # replacement con un certo numero di zeri se necessario
```

```python
'this is a float %f' % 15.3456   # replacement di un float object
```

```python
'this is a float %.2f' % 15.3456   # replacement con un numero fisso di decimali
```

```python
'this is a float %8f' % 15.3456   # replacement con un numero fisso di caratteri
```

#### nuova formattazione

```python
'this is an integer {:d}' .format(15)   # replacement di un int object
```

```python
'this is an integer {:4d}' .format(15)  # replacement con un numero fisso di caratteri
```

```python
'this is an integer {:04d}' .format(15)  # replacement con un certo numero di zeri se necessario
```

```python
'this is a float {:f}' .format(15.3456)   # replacement di un float object
```

```python
'this is a float {:.2f}' .format(15.3456) # replacement con un numero fisso di decimali
```

```python
'this is a float {:8f}' .format(15.3456)  # replacement con un numero fisso di caratteri
```

```python
'this is a float {:8.2f}' .format(15.3456) # replacement con un numero fisso di caratteri e decimali
```

```python
'this is a string {:s}' .format('Python')   # replacement di uno str object
```

```python
'this is a string {:10s}' .format('Python')    # replacement con un numero fisso di caratteri
```

#### la sostituzione di stringhe è particolarmente utile nel contesto di operazioni di stampe multiple in cui il dato da stampare è aggiornato durante un ciclo while:

```python
i = 0
while i < 4:
    print('il numero è %d' % i)
    i = i + 1
```

```python
i = 0
while i < 4:
    print('il numero è {:d}' .format(i))
    i = i + 1
```

```python
from itertools import zip_longest
```

```python
numbers = [1, 2, 3]
letters = ['a', 'b', 'c']
```

```python
longest = range(5)
```

```python
longest
```

```python
zipped = zip_longest(numbers, letters, longest, fillvalue='?')
list(zipped)
```

```python
l = [0, 1, 2, 3, 4]
```

```python
zipped = zip_longest(numbers, letters, l, fillvalue='?')
list(zipped)
```
