### Presentare il codice Python che implementa una pila mediante liste concatenate. Si richiede, in particolare, la definizione della classe Node (per gli elementi della lista), della classe Stack (la pila) e i metodi Push (inserimento di un elemento) e Pop (estrazione di un elemento).

```python
class Node:
    data=''
    next=None

    def __init__(self, data):
        self.data=data

class Stack:
    __top=None
    __size=0

def push(self, element):
    newNode=Node(element)
    if(self.__top==None):
        self.__top=newNode
    else:
        newNode.next=self.__top
        self.__top=newNode
    self.__size+=1


def pop(self):
    if(self.__top==None):
        return None
    else:
        p=self.__top
        self.__top=self.__top.next
        p.next=None
        self.__size-=1
        return p
```

### Illustrare nel dettaglio l'algoritmo "Hill Climbing" nella sua versione First-Choice Hill Climbing, presentando anche il codice Python opportunamente commentato per il problema di ricerca di una stringa.

L'algoritmo Hill Climbing fa parte della categoria della ricerca locale. Ci sono problemi in cui lo stato obiettivo contiene tutte le informazioni rilevanti per la soluzione, infatti in questi casi il cammino verso l'obiettivo è irrilevante. Gli algoritmi di ricerca locale si muovono sulla superficie cercando i picchi più alti. Infatti il termine Hill Climbing deriva proprio da questo, infatti l'algoritmo segue sempre le salite più rapide. La versione che implementiamo in Python si chiama First-Choice Hill Climbing, esso dato lo stato corrente, sceglie casualmente un successore, e passa al nuovo stato se la valutazione di tale successore è migliore di quella dello stato corrente.


Lo utilizziamo per risolvere il problema della ricerca di una stringa, data una stringa iniziale di lunghezza n e fissata una stringa obiettivo di lunghezza n, trovare la stringa obiettivo a partire da quella iniziale.

```python
import random
import string
import numpy
import math
```

Funzione che genera random una stringa

```python
def generate_random_string(length = 8):
    return [random.choice(string.printable) for _ in range(length)] # String printable ci restituisce i caratteri disponibili
```

```python
def evaluate(solution): # Valuta la distanza tra la stringa corrente e la stringa obiettivo
    target = list('Roma Tre')
    diff = 0
    for i in range(len(target)):
        s = solution[i]
        t = target[i]
        delta = abs(ord(s) - ord(t))  # ord serve a convertire il carattere nel codice unicode
        diff = diff + delta # Distanza delle stringhe calcolata come somma della distanza tra ciascuno dei caratteri
    return diff
```

Funzione che sceglie random uno stato successore

```python
def tweak(solution):
    index = random.randint(0, len(solution) - 1) # Scelgo random un indice
    solution[index] = random.choice(string.printable) # Scelgo random un carattere da sostituire al carattere che si trova nella posizione dell'indice precedentemente estratto
```

Algoritmo

```python
def ricerca_stringa(stringa_iniziale):
    
    current_state = stringa_iniziale
    current_eval = evaluate(current_state)
    while True:
        print('Punteggio attuale:', current_eval, 'Stringa:', "".join(current_state)) # Questo join unisce current_state e ce lo resituisce senza spazi o separatori
    
        if current_eval == 0: # Siamo arrivati alla soluzione
            break
              
        next_state = list(current_state)       #  scelta random
        tweak(next_state)                      #  del successore
    
        next_state_eval = evaluate(next_state) # valutazione del successore scelto
        
        if next_state_eval < current_eval:     # se next è migliore ...
            current_state = next_state         # ... esso diventa
            current_eval = next_state_eval     # il nuovo nodo corrente]
                 
```

Ci sono problemi in cui lo stato obiettivo contiene tutte le informazioni rilevanti per la soluzione, infatti in questi casi il cammino verso l'obiettivo è irrilevante. Gli algoritmi di ricerca locale si muovono sulla superficie cercando i picchi più alti.
L'algoritmo di Simulated Annealing prende il nome dall'analogia con il processo di ricottura in metallurgia che consiste nel ridurre gradualmente la temperatura di un materiale in modo da permettere alla sua struttura cristallina di raggiungere uno stato ad energia minima. Il Simulated Annealing comincia con valori alti di T comportandosi in modo simile ad una pura random search. Poi decrementa gradualmente il valore di T. Verso la fine dell'esecuzione i valori di T sono molto piccoli e perciò l'algoritmo si comporta in modo simile ad un normale hill-climber. Inoltre l'algoritmo accetta sempre nuovi punti se essi hanno una valutazione migliore del punto corrente. La funzione tweak viene utilizzata per scegliere random uno stato successore. La funzione inizializza serve per scegliere casualmente lo stato di partenza. La funzione energia è la funzione di valutazione.

Import

```python
import random
import numpy as np
import math
```

Impostazione parametri

```python
# Impostazione parametri

TEMPERATURA_INIZIALE = 30
TEMPERATURA_FINALE = 0.2
ALFA = 0.99 # Coefficiente per diminuire la temperatura T
STEPS_PER_CHANGE = 100 # Numero di tentativi che l algoritmo fa per una determinata temperatura

DIMENSIONE = 8   # dimensione dei lati della scacchiera N x N (dove N è la DIMENSIONE)
```

Funzione Tweak (Sceglie random uno stato successore) (l'encoding ha come indice la colonna e come valore la riga)

```python
def tweak(sol): 
    
    sol_copy = np.copy(sol)
    
    # scegli random due colonne distinte
    x = random.randint(0,DIMENSIONE-1) 
    y = random.randint(0,DIMENSIONE-1)
    while x==y:
        y = random.randint(0,DIMENSIONE-1)
        
    # scambia le due colonne
    temp = sol_copy[y]
    sol_copy[y] = sol_copy[x] 
    sol_copy[x] = temp
    
    return sol_copy
```

Funzione inizializza (Sceglie random lo stato di partenza)

```python
def inizializza(sol): 
    
    # shake shake shake
    for c in range(0,DIMENSIONE-1):
        sol = tweak(sol)
    return sol
```

Funzione energia (Sarebbe la funzione di valutazione e calcola il numero di attacchi/conflitti)

```python
def energia(stato): # Stato sarebbe l encoding
    
    # definizione della scacchiera N x N
    board  = [[0] * DIMENSIONE for i in range(DIMENSIONE)] # è una lista di liste
    
    # inserimento delle regine ('Q') nelle loro posizioni sulla scacchiera
    for i in range(0,DIMENSIONE):
        board[stato[i]][i] ='Q'
        
    # spostamenti possibili sulla scacchiera
    dx = [-1,1,-1,1]
    dy = [-1,1,1,-1]
    
    # inizializzazione numero di attacchi (diretti o indiretti)
    conflitti = 0

    for i in range(0,DIMENSIONE):   # Per ogni regina verifico se ci sono conflitti da tutte le direzioni     
        x = stato[i] # Riga 
        y = i # Colonna
         
        # verifica attacchi sulle diagonali       
        for j in range(0,4):
            tempx = x
            tempy = y
            while (True):
                tempx = tempx + dx[j]           # spostamento sull'asse x
                tempy = tempy + dy[j]           # spostamento sull'asse y
                if ((tempx < 0) or 
                    (tempx >= DIMENSIONE) or # Ci si continua a spostare sulla diagonale finchè non si esce dalla scacchiera
                    (tempy < 0) or           # Questo perchè dobbiamo controllare tutta la diagonale
                    (tempy >= DIMENSIONE)):
                    break                       # si esce se lo spostamento va fuori la scacchiera
                if (board[tempx][tempy]=='Q'):
                    conflitti = conflitti + 1   # aggiornamento numero di attacchi
    return conflitti
```

Algoritmo

```python
def simulated_annealing():
    
    # impostazione dello stato iniziale
    current = inizializza(range(0,DIMENSIONE))
    current_energy = energia(current)
    
    # inizializzazione best
    best = current
    best_energy = current_energy

    temperature = TEMPERATURA_INIZIALE

    while (temperature > TEMPERATURA_FINALE and best_energy != 0):
        print("TEMPERATURA: "),    
        print(temperature)
        for step in range(0,STEPS_PER_CHANGE):
            useNew = False # Può essere True o False e ci dice se accettare o no il nuovo stato
            next_state = tweak(current)                # scelta random dello stato successore nel neighbourhood
            next_state_energy = energia(next_state)    # valutazione dell'energia dello stato successore
    
            if (next_state_energy < current_energy):   # se il successore è migliore lo accettiamo
                useNew = True
            else:
                delta = next_state_energy - current_energy
                metropolis = math.exp(-delta/temperature)  # calcolo probabilità di accettazione (e^-deltaE/T) dove deltaE rappresenta quanto il nuovo stato è peggiore
                test = random.random() # La probabilità viene confrontata con un numero random che va da 0 a 1
                if (test < metropolis):                # se il numero random è minore della probabilità ...
                    useNew = True                      # ... accettiamo il nuovo stato 
                    
            # se abbiamo deciso di accettare il nuovo stato:   
            if (useNew):
                # impostalo come stato ed energia correnti
                current = next_state
                current_energy = next_state_energy
            
                # se è anche il migliore segna il record
                if (current_energy < best_energy):
                    best = current
                    best_energy = current_energy

        print("best_energy="),
        print(best_energy)  
        
        # diminuisci la temperatura, senza mai arrivare a zero
        temperature = temperature * ALFA
        
    return(best)
```
