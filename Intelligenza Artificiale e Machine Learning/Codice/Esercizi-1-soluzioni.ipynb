## Fattoriale

```python
def fatt(n):
    if n <= 1:
        return 1
    else:
        return n * fatt(n-1)
```

```python
fatt(10)
```

```python
# versione poco efficiente, perchè ogni volta ricomincia da capo
def fatt_list(n):         # lista fattoriali da 1 a n per n >= 1
    sequenza = [] # imposto a lista vuota
    x = 1
    while x <= n:
        sequenza.append(fatt(x))  # fatt(x) ricomincia sempre dall'inizio
        x = x + 1
    return(sequenza)
```

```python
fatt_list(10)
```

```python
def fattoriale(n):        # lista fattoriali da 1 a n per n >= 1
    if n <= 1:
        return [1] 
    else:
        sequenza = fattoriale(n-1) # qui supponiamo che metta il fattoriale di n-1
        sequenza.append(sequenza[-1] * n) # ora qui devo solamente appendere il fattoriale di n, quindi n x l ultimo elemento della sequenza precedente
        return sequenza
```

```python
fattoriale(10)
```

```python
def fattoriale_iter(n):   # lista fattoriali da 1 a n per n >= 1
    sequenza = [1]
    if n <= 1:
        return [1] 
    else: 
        for i in range(2, n+1):
            sequenza.append(sequenza[-1] * i)
        return sequenza
```

```python
fattoriale_iter(10)
```

## Sequenza di Fibonacci

```python
def fibonacci(n):   # sequenza dei primi n+1 elementi
    if n == 0:
        return [0]
    elif n == 1:
        return [0, 1] 
    else:
        sequenza = fibonacci(n-1)
        sequenza.append(sequenza[-1] + sequenza[-2]) # aggiungo l ultimo elemento che è la somma degli ultimi due elementi
        return sequenza
```

```python
def fibonacci_iter(n):
    if n == 0:
        return [0]
    sequenza = [0, 1]
    for i in range(2, n+1):
        sequenza.append(sequenza[-1] + sequenza[-2])
    return sequenza
```

```python
fibonacci_iter(1)
```

```python
fibonacci_iter(10)
```

```python
def fib(n):  # calcola l'elemento Fn della sequenza in modo ricorsivo
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2) # Questa è quella più inefficiente perchè mi capita di richiamare sugli stessi numeri
```

```python
fib(10)
```

```python
def fib_iter(n):   # calcola l'elemento Fn della sequenza in modo iterativo
    if n == 0:
        return n
    x, y = (0, 1)
    for i in range(1, n):
        x, y = (y, x + y)
    return y
```

```python
fib_iter(10)
```

```python
def fib(n):
    """assumiamo n intero >= 0"""
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2)
```

### Ci sono due approcci alla "dynamic programming":

#### Memoization: 
Risolve il problema originario *top down*. Scompone il problema originario in sottoproblemi, scompone i vari sottoproblemi in ulteriori sottoproblemi e così via. Ogni volta che un sottoproblema è risolto, si memorizza il risultato in una tabella. Ogni volta che è necessario risolvere lo stesso sottoproblema, si cerca prima la soluzione nella tabella.

#### Tabular:
Risolve il problema originario *bottom up*. Si parte dai problemi più piccoli e si memorizzano le loro soluzioni in una tabella. Quindi si combinano le soluzioni a questi problemi per risolvere i successivi problemi più piccoli e si memorizzano quelle risposte nella tabella.

### Fibonacci con "*memoization*"
Complessità: *O(n)*

```python
def fib_memo(n, memo = None):
    """assumiamo n intero >= 0"""
    if memo == None: # tabella in cui memorizziamo
        memo = {}
    if n <= 1:
        return n
    try:
        return memo[n] # il valore associato alla chiave n nella tabella memo, ogni volta memorizza il fattoriale di m
    except:
        risultato = fib_memo(n-1, memo) + fib_memo(n-2, memo)
        memo[n] = risultato
        return risultato
```

```python
fib_memo(10)
```

```python
%time fib_memo(35)
```

```python
%time fib_memo(40)
```

```python
fib_memo(40)
```

```python
fib_memo(42)
```

```python
fib_memo(50)
```

```python
fib_memo(130)
```

```python
memo = {}
def fib(n):
    if n <= 2:
        return 1
    if n not in memo:
        memo[n] = fib(n-1) + fib(n-2)
    return memo[n]
```

## Fibonacci con "*tabular*"
Complessità: *O(n)*

```python
def fib_tab(n):
    """assumiamo n intero >= 0"""
    tab = [0, 1]
    for i in range(2, n+1):
        tab.append(tab[i-1] + tab[i-2])
    return tab[n]
```

```python
fib_tab(10)
```

```python
%time fib_tab(130)
```

```python
%time fib_tab(130)
```
