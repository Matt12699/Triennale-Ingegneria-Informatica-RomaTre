### Domanda n. 4

```python
def cost(x): # Qualsiasi cosa gli passi restituisce 1
    return 1.

def lin(x): # lineare
    return x

def quad(x): # quadrato
    return x**2

def cubo(x): # cubo
    return x**3
```

```python
lista_funz = [cost, lin, quad, cubo] # posso definire una lista di funzioni
```

```python
lista_funz[3](4) 
```
